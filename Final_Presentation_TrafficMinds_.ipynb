## 1. Introduction

This project explores a **Computer Vision solution** to identify traffic light states from images.

### Problem Context
Traffic lights can sometimes be difficult to recognize due to:

- glare from sunlight
- bad weather conditions
- low lighting environments
- unusual camera angles
- color vision deficiencies

### Goal
The goal of this project is to **train an AI system capable of accurately identifying traffic light states from images**, even under challenging real-world conditions.

## 2. Dataset

For this project we used the **LISA Traffic Light Dataset**, a real-world dataset containing images of traffic lights captured from driving environments.

### Dataset Characteristics

- Real-world traffic scenes
- Different lighting conditions
- Multiple environmental scenarios
- Traffic lights captured from moving vehicles

### Classes predicted by the model

The models classify each image into one of the following categories:

- **Red**
- **Yellow**
- **Green**
- **Unknown**

The goal of the model is to analyze a traffic light image and correctly predict its state.

## Why This Project Matters

Traffic light recognition is a **real-world safety issue**.

If a driver misinterprets a traffic light, even for a few seconds, it can lead to **accidents or dangerous decisions**.

This becomes more challenging in conditions like **rain, low light, or strong sunlight glare**.

This project shows how **Computer Vision** can help detect traffic light colors more reliably than human perception alone.

In real-world applications, this system could support:

- driver assistance systems  
- traffic monitoring  
- accessibility for color-blind users  

Although this is a prototype, it demonstrates how AI can help create **safer and more reliable transportation systems**.

## 3. Data Preparation

Before training the models, several preprocessing steps were applied to prepare the dataset for deep learning.

### Preprocessing Steps

- **Image resizing** to ensure all images have the same dimensions
- **Normalization** to stabilize model training
- **Data transformations** to improve generalization
- Preparation of the dataset for **transfer learning**

### Class Imbalance Consideration

The dataset is not perfectly balanced across classes.  
To address this, we evaluated model performance using **macro-F1 score**, which gives equal importance to all classes, including minority classes.

## 4. Models Tested

To solve the traffic light classification problem, we evaluated three different **transfer learning architectures**.

Transfer learning allows us to use models that were previously trained on large image datasets and adapt them to our specific task.

### Models Evaluated

**ResNet18**
- A well-known convolutional neural network architecture
- Often used as a strong baseline for image classification tasks

**EfficientNet-B0**
- A modern architecture designed to achieve strong performance with efficient computation
- Known for balancing model accuracy and efficiency

**MobileNetV3-Small**
- A lightweight model optimized for mobile and edge devices
- Designed to run efficiently on systems with limited computational resources

All three models were trained and evaluated using the same dataset and preprocessing pipeline.

## 5. Model Evaluation

To compare the performance of the models, we evaluated them using several metrics commonly used in machine learning classification tasks.

### Evaluation Metrics

- **Accuracy** – percentage of correct predictions
- **Precision** – how many predicted positives are actually correct
- **Recall** – how many real positives were correctly detected
- **F1-Score** – balance between precision and recall
- **Macro-F1** – evaluates performance equally across all classes
- **Confusion Matrix** – shows where the model makes classification mistakes

Using multiple metrics is important because the dataset is not perfectly balanced across classes.

The confusion matrix also helps us understand which traffic light states the model may confuse under difficult conditions.

## 6. Model Comparison

After training the three models, we compared their performance using the test dataset.

| Model | Test Accuracy | Macro-F1 |
|------|--------------|----------|
| ResNet18 | 0.9752 | 0.9672 |
| EfficientNet-B0 | **0.9850** | **0.9795** |
| MobileNetV3-Small | 0.9832 | 0.9762 |

### Key Observations

- All three models performed well on the dataset.
- **EfficientNet-B0 achieved the highest accuracy and macro-F1 score.**
- MobileNetV3-Small also performed strongly while remaining lightweight.
- ResNet18 served as a stable baseline model.

These results indicate that **EfficientNet-B0 provides the most balanced and reliable performance across all traffic light classes.**

## 7. Model Selection

Based on the comparison results, **EfficientNet-B0** was selected as the final model for this project.

### Reasons for Selection

- Highest **test accuracy**
- Highest **macro-F1 score**
- More **balanced performance across all classes**
- Lower risk of critical misclassifications

EfficientNet-B0 demonstrated the most **consistent and reliable performance**, making it the best candidate for a real-world traffic light recognition system.
